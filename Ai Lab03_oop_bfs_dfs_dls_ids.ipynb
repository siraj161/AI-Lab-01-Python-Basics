## 7. Breadth-First Search (BFS)
BFS explores level by level using a FIFO queue. With equal edge costs, it finds a path with the fewest edges. Visit order and goal path are different.

### Worked examples

```python
from collections import deque

def bfs_order(graph, start):
    queue, visited, order = deque([start]), {start}, []
    while queue:
        node = queue.popleft()
        order.append(node)
        for nxt in graph.get(node, []):
            if nxt not in visited:
                visited.add(nxt); queue.append(nxt)
    return order

def bfs(graph, start, goal):
    queue, visited, parent = deque([start]), {start}, {start: None}
    while queue:
        node = queue.popleft()
        if node == goal:
            path = []
            while node is not None:
                path.append(node); node = parent[node]
            return path[::-1]
        for nxt in graph.get(node, []):
            if nxt not in visited:
                visited.add(nxt); parent[nxt] = node; queue.append(nxt)
    return None

teaching_graph = {"A": ["B", "C"], "B": ["D", "E"], "C": ["F", "G"],
                  "D": [], "E": [], "F": [], "G": []}
print("BFS order:", bfs_order(teaching_graph, "A"))
print("Path to G:", bfs(teaching_graph, "A", "G"))
```

## 8. Depth-First Search (DFS)
DFS follows one branch deeply before backtracking. It uses a LIFO stack. To visit the leftmost neighbor first, push neighbors in reverse order.

### Worked examples

```python
def dfs_order(graph, start):
    stack, visited, order = [start], set(), []
    while stack:
        node = stack.pop()
        if node in visited: continue
        visited.add(node); order.append(node)
        for nxt in reversed(graph.get(node, [])):
            if nxt not in visited: stack.append(nxt)
    return order

def dfs(graph, start, goal):
    stack, visited = [(start, [start])], set()
    while stack:
        node, path = stack.pop()
        if node in visited: continue
        visited.add(node)
        if node == goal: return path
        for nxt in reversed(graph.get(node, [])):
            if nxt not in visited: stack.append((nxt, path + [nxt]))
    return None

print("DFS order:", dfs_order(teaching_graph, "A"))
print("DFS path to F:", dfs(teaching_graph, "A", "F"))
```

## 9. Depth-Limited Search (DLS)
DLS is DFS with a maximum depth. The start node is depth 0. A node at the limit can be checked, but its children are not expanded.

### Worked examples

```python
def dls(graph, start, goal, limit):
    def visit(node, depth, path):
        if node == goal: return path
        if depth == limit: return None
        for nxt in graph.get(node, []):
            if nxt not in path:
                result = visit(nxt, depth + 1, path + [nxt])
                if result is not None: return result
        return None
    return visit(start, 0, [start])

print("Limit 1:", dls(teaching_graph, "A", "G", 1))
print("Limit 2:", dls(teaching_graph, "A", "G", 2))
```

## 10. Iterative Deepening Search (IDS)
IDS repeats DLS with limits 0, 1, 2, ... until the goal is found or the maximum depth is reached.

### Worked examples

```python
def dls_path(graph, node, goal, limit, path):
    if node == goal: return path
    if limit == 0: return None
    for nxt in graph.get(node, []):
        if nxt not in path:
            result = dls_path(graph, nxt, goal, limit - 1, path + [nxt])
            if result is not None: return result
    return None

def ids(graph, start, goal, max_depth):
    for limit in range(max_depth + 1):
        print("Trying depth:", limit)
        result = dls_path(graph, start, goal, limit, [start])
        if result is not None: return result
    return None

print("IDS result:", ids(teaching_graph, "A", "G", 3))
```

## 11. BFS / DFS / DLS / IDS Comparison

| Algorithm | Main structure | Main idea |
|---|---|---|
| BFS | FIFO queue | Explore level by level; fewest edges when costs are equal |
| DFS | LIFO stack or recursion | Go deep, then backtrack |
| DLS | Depth-first with a limit | Do not search deeper than the chosen limit |
| IDS | Repeated DLS | Try limits 0, 1, 2, ... until found |

**Remember:** Traversal order lists nodes visited. A goal path lists only the nodes on one route from start to goal.

### Challenge 8: GraphSearch class with graph attribute and bfs(start, goal).

```python
class GraphSearch:
    def __init__(self, graph): self.graph = graph
    def bfs(self, start, goal):
        queue, visited, parent = deque([start]), {start}, {start: None}
        while queue:
            node = queue.popleft()
            if node == goal:
                path = []
                while node is not None:
                    path.append(node); node = parent[node]
                return path[::-1]
            for nxt in self.graph.get(node, []):
                if nxt not in visited:
                    visited.add(nxt); parent[nxt] = node; queue.append(nxt)
        return None

searcher = GraphSearch(teaching_graph)
print(searcher.bfs("A", "G"))
```
