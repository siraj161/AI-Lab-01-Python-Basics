```python
positive = 0
negative = 0
zero = 0

for i in range(10):
    number = float(input("Enter a number: "))

    if number > 0:
        positive = positive + 1
    elif number < 0:
        negative = negative + 1
    else:
        zero = zero + 1

print("Positive:", positive)
print("Negative:", negative)
print("Zero:", zero)
```

```python
number = int(input("Enter a number: "))

if number < 2:
    print("Not prime")
else:
    is_prime = True

    for i in range(2, number):
        if number % i == 0:
            is_prime = False
            break

    if is_prime:
        print("Prime")
    else:
        print("Not prime")
```
